# Import Data

```python
import numpy as np
import pandas as pd

# Import covariance matrices n=2
python_cov_2 = pd.read_csv("FYP/python_cov_2.csv", index_col=0)
julia_cov_2 = np.loadtxt("FYP/julia_cov_2.csv", delimiter=",")

# Import correlation matrices n=2
python_corr_2 = pd.read_csv("FYP/python_corr_2.csv", index_col=0)
julia_corr_2 = np.loadtxt("FYP/julia_corr_2.csv", delimiter=",")

# Import covariance matrices n=5
python_cov_5 = pd.read_csv("FYP/python_cov_5.csv", index_col=0)
julia_cov_5 = np.loadtxt("FYP/julia_cov_5.csv", delimiter=",")

# Import correlation matrices n=5
python_corr_5 = pd.read_csv("FYP/python_corr_5.csv", index_col=0)
julia_corr_5 = np.loadtxt("FYP/julia_corr_5.csv", delimiter=",")

# Import covariance matrices n=10
python_cov_10 = pd.read_csv("FYP/python_cov_10.csv", index_col=0)
julia_cov_10 = np.loadtxt("FYP/julia_cov_10.csv", delimiter=",")

# Import correlation matrices n=10
python_corr_10 = pd.read_csv("FYP/python_corr_10.csv", index_col=0)
julia_corr_10 = np.loadtxt("FYP/julia_corr_10.csv", delimiter=",")

# Import covariance matrices n=50
python_cov_50 = pd.read_csv("FYP/python_cov_50.csv", index_col=0)
julia_cov_50 = np.loadtxt("FYP/julia_cov_50.csv", delimiter=",")

# Import correlation matrices n=50
python_corr_50 = pd.read_csv("FYP/python_corr_50.csv", index_col=0)
julia_corr_50 = np.loadtxt("FYP/julia_corr_50.csv", delimiter=",")
```

# n = 2

```python
cov_diff_2 = np.abs(python_cov_2.values - julia_cov_2)
corr_diff_2 = np.abs(python_corr_2.values - julia_corr_2)

print("Covariance differences:")
print(cov_diff_2)

print("Correlation differences:")
print(corr_diff_2)
```

```python
print("Maximum covariance difference:", np.max(cov_diff_2))
print("Maximum correlation difference:", np.max(corr_diff_2))
```

```python
print("Covariance agreement:",
      np.allclose(python_cov_2, julia_cov_2, rtol=1e-8, atol=1e-12))

print("Correlation agreement:",
      np.allclose(python_corr_2, julia_corr_2, rtol=1e-8, atol=1e-12))
```

```python
relative_diff = (
    abs(python_corr_2.iloc[0, 1] - julia_corr_2[0, 1])
    / abs(python_corr_2.iloc[0, 1])
)

print("Relative correlation difference:", relative_diff)
print("Relative correlation difference (%):", relative_diff * 100)
```

## n = 5 (changing Julia Matrices First)

```python
print(python_corr_5)
```

```python
print(julia_corr_5)
```

This shows that Python and Julia exported the data differently due to the very large maximum correlation differences. Python exported the stocks in alphabetical order, Julia exported them exactly as is.

```python
julia_stocks_5 = ["AAPL", "TSLA", "MSFT", "KO", "JNJ"]

julia_corr_5 = pd.DataFrame(
    julia_corr_5,
    index=julia_stocks_5,
    columns=julia_stocks_5
)

julia_corr_5 = julia_corr_5.loc[
    python_corr_5.index,
    python_corr_5.columns
]

print(julia_corr_5)
```

```python
corr_diff_5 = np.abs(
    python_corr_5 - julia_corr_5
)

print("Maximum correlation difference:",
      corr_diff_5.to_numpy().max())

print("Numerical agreement:",
      np.allclose(python_corr_5, julia_corr_5,
                  rtol=1e-8, atol=1e-12))
```

```python
julia_cov_5 = pd.DataFrame(
    julia_cov_5,
    index=julia_stocks_5,
    columns=julia_stocks_5
)

julia_cov_5 = julia_cov_5.loc[
    python_cov_5.index,
    python_cov_5.columns
]

print("Maximum covariance difference:",
      np.abs(python_cov_5 - julia_cov_5).to_numpy().max())
```
